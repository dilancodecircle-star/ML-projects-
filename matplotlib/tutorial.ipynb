```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0 ,10 , 100)
print(x)
```

```python
y = np.sin(x)
z= np.cos(x)
```

```python
print(y)
```

```python
print(z)
```

```python
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,z)
plt.show()
```

```python
t = np.tan(x)
plt.plot(x , t)
plt.show()
```

```python
plt.plot(x ,y , color = 'red')
plt.xlabel('angle')
plt.ylabel('sin(x)')
plt.title('sin wave')
plt.show()
```

```python
x = np.linspace(-10 , 10 , 20)
print(x)
y = x**2
plt.plot(x , y , 'r+')
plt.xlabel('x')
plt.ylabel('y')
plt.title('parabola')
plt.show()
```

```python
plt.plot(x , y , 'g.')
plt.xlabel('x')
plt.ylabel('y')
plt.title('parabola')
plt.show()
```

```python
plt.plot(x , y , 'rx')
plt.xlabel('x')
plt.ylabel('y')
plt.title('parabola')
plt.show()
```
