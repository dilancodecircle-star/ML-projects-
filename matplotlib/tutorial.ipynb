```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0 ,10 , 100)
print(x)
```

```python
y = np.sin(x)
z= np.cos(x)
```

```python
print(y)
```

```python
print(z)
```

```python
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,z)
plt.show()
```

```python
t = np.tan(x)
plt.plot(x , t)
plt.show()
```

```python
plt.plot(x ,y , color = 'red')
plt.xlabel('angle')
plt.ylabel('sin(x)')
plt.title('sin wave')
plt.show()
```

```python
x = np.linspace(-10 , 10 , 20)
print(x)
y = x**2
plt.plot(x , y , 'r+')
plt.xlabel('x')
plt.ylabel('y')
plt.title('parabola')
plt.show()
```

```python
plt.plot(x , y , 'g.')
plt.xlabel('x')
plt.ylabel('y')
plt.title('parabola')
plt.show()
```

```python
plt.plot(x , y , 'rx')
plt.xlabel('x')
plt.ylabel('y')
plt.title('parabola')
plt.show()
```

```python
x = np.linspace(-5 ,5 , 50)
plt.plot(x , np.sin(x) , 'gx')
plt.plot(x , np.cos(x) , 'r--')
plt.show()
```

```python
#Bar plot
```

```python
fig = plt.figure()
ax= fig.add_axes([0,0 ,1,1])
language = ["English" , "French" , "Spanish" , "Latin" , "German"]
peoples = [100 , 10 , 50, 54 , 60]
ax.bar(language , peoples)
plt.xlabel("langages")
plt.ylabel("peoples")
```

```python
fig1= plt.figure()
ax =fig1.add_axes([0,0,2,2])
language = ["English" , "French" , "Spanish" , "Latin" , "German"]
peoples = [100 , 10 , 50, 54 , 60]
ax.pie(peoples , labels=language )
plt.show()

plt.pie(peoples , labels=language , autopct= '%1.1f%%')
plt.show()

```

```python
x = np.linspace(0 , 10 , 30)
y = np.sin(x)
z = np.cos(x)
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])
ax.scatter(x,y , color = 'b')
ax.scatter(x,z , color = 'green' )
```

```python

ax = plt.axes(projection = '3d')
x = 20*np.random.random(100)
y = np.sin(x)
z = np.cos(x)
ax.scatter(x ,y, z ,c = z , cmap = "Greens" )
plt.show()
```
