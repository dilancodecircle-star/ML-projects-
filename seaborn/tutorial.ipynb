```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = 20*np.linspace(-10 , 10 , 20)
y = np.sin(x)
plt.plot(x ,y)
plt.show()
```

```python
ax = plt.axes(projection = '3d')
z = np.cos(x)
ax.scatter(x , y, z , c =z , cmap = 'Greens')
```

```python
iris = sns.load_dataset('iris')
iris.head()
```

```python
sns.scatterplot(x = 'sepal_length' , y ='petal_width' , hue = 'species'  , data = iris )
```

```python
dots = sns.load_dataset('dots')
dots.head()
```

```python
sns.relplot(x = 'time' , y='firing_rate' , col ='align' , hue = 'choice' , size = 'coherence' , style ='choice' , data = dots , kind = 'line')   
```

```python
fmri =sns.load_dataset('fmri')
fmri.head()
```

```python
sns.relplot(x= 'timepoint' , y= 'signal' , hue ='event' ,style = 'event', data = fmri , kind = 'line' )
```

```python
tip = sns.load_dataset('tips')
tip.head()
```

```python
sns.lmplot(x='total_bill' , y ='tip' , col = 'time' ,  hue ='smoker' ,data=tip )
```
