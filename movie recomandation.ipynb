```python
import ast
```

```python
def convert(obj):
    l=[]
    for i in ast.literal_eval(obj):
        l.append(i['name'])
    return l
```

```python
movie_data['genres'] = movie_data['genres'].apply(convert)
```

```python
movie_data['keywords'] = movie_data['keywords'].apply(convert)
```

```python
movie_data.head()
```

```python
movie_data['cast'] = movie_data['cast'].apply(convert)
```

```python
movie_data.head()
```

```python
def convert3(obj):
    l =[]
    counter=0
    for i in ast.literal_eval(obj):
        if counter!=3:
            l.append(i['name'])
            counter+=1
        else:
            break
        return l
```

```python
movie_data['cast'] = movie_data['cast'].apply(convert3)
```

```python
movie_data.head()
```

```python
def fetche_director(obj):
    l=[]
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            l.append(i['name'])
        return l
```

```python
movie_data['crew'] = movie_data['crew'].apply(fetche_director)
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y=[]
    for i in text.split():
        y.append(ps.stem(i))

    return " ".join(y)
```

```python
new_data['tags'] = new_data['tags'].apply(stem)
```
